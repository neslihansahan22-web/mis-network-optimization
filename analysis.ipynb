# Executive Summary
This analysis investigates the most cost-efficient transportation routes within a simulated Istanbul logistics network. By applying graph theory, we aim to transform raw shipping data into an optimized operational strategy.

```python
#Kütüphaneler
import networkx as nx
import pandas as pd
import matplotlib.pyplot as plt
import os

# Görselleştirme ayarları
plt.rcParams['figure.figsize'] = [12, 8]
plt.rcParams['figure.dpi'] = 100
```

```python
# Veriyi oku
data_path = os.path.join('..', 'data', 'network_data.csv')
df = pd.read_csv(data_path)

print("### Network Data Overview ###")
display(df)

print(f"\nTotal Connections: {len(df)}")
print(f"Average Shipping Cost: {df['weight'].mean():.2f} TL")
```

## 2. Graph Construction
We model each location as a **Node** and each route as an **Edge**. The shipping cost (TL) is assigned as the **Weight** of each edge.

```python
# Algoritma Uyg
# Çizgeyi oluştur
G = nx.Graph()
for _, row in df.iterrows():
    G.add_edge(row['source'], row['target'], weight=row['weight'])

# En kısa yolu hesapla
source_node = "Lojistik_Merkezi"
target_node = "Tuzla_Depo"

optimal_path = nx.shortest_path(G, source=source_node, target=target_node, weight='weight')
min_cost = nx.shortest_path_length(G, source=source_node, target=target_node, weight='weight')

print(f"🎯 Analysis Complete!")
print(f"Optimal Strategy: {' ➔ '.join(optimal_path)}")
print(f"Minimum Operational Cost: {min_cost} TL")
```

```python
# Görsel Analiz
# Görselleştirme
pos = nx.spring_layout(G, seed=42) # Sabit yerleşim için seed ekledik

# Düğümleri çiz
nx.draw_networkx_nodes(G, pos, node_size=2500, node_color='skyblue')
nx.draw_networkx_labels(G, pos, font_size=10, font_family='sans-serif')

# Kenarları çiz (Yolu vurgulayalım)
path_edges = list(zip(optimal_path, optimal_path[1:]))
nx.draw_networkx_edges(G, pos, width=1, alpha=0.5)
nx.draw_networkx_edges(G, pos, edgelist=path_edges, width=3, edge_color='red')

# Ağırlıkları yaz
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.title(f"Optimized Logistics Route (Red Path)\nCost: {min_cost} TL", size=15)
plt.axis('off')
plt.show()
```

## 3. Conclusion & Recommendations
Based on the Dijkstra algorithm results:
1. **Efficiency:** The route via **Uskudar** and **Umraniye** is the most cost-effective path.
2. **Impact:** Automating this decision process can reduce logistics expenses by eliminating human error in route selection.
